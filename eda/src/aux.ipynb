```python
import pandas as pd

# 1. Cargar el archivo CSV
df = pd.read_csv("dataset_normalized.csv")

# 2. Seleccionar solo las columnas que NO son de tipo booleano
# El parámetro exclude=['bool'] se encarga de "fumarse" las de True/False
df_sin_booleanos = df.select_dtypes(exclude=['bool'])

# Opcional: ver cuántas columnas quedaron
print(f"Columnas originales: {df.shape[1]}")
print(f"Columnas después de eliminar booleanos: {df_sin_booleanos.shape[1]}")

# 3. Guardar el nuevo DataFrame en un archivo CSV
df_sin_booleanos.to_csv("dataset_sin_booleanos.csv", index=False)
```
